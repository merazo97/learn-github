```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Creamos un dataset simulando un registro de incidentes detectados por malware
data = {
    'timestamp': pd.date_range(start='2024-01-01', periods=15, freq='h'),
    
    'tipo de incidente': ['Hardware','Software','Red','Base de datos','Inquiry','Software','Base de datos','Base de datos','Hardware','Software',
                          'Red','Red','Inquiry','Red','Hardware'],
    'ip_afectadas': ['192.168.1.' + str(i) for i in range(1, 16)],
    'puertos instrusivos': [80, 445, 3389, 23, 80, 445, 8080, 3389, 80, 445, 
                            80, 445, 23, 3389, 445],
    'protocolos inseguros vulnerables': ['HTTP', 'HTTP', 'TELNET', 'RDP', 'FTP', 'HTTP', 'FTP', 'TELNET', 'HTTP', 'RDP',
                                         'SMTP','HTTP', 'TELNET', 'RDP', 'FTP'],
    'bytes_enviados': np.random.randint(1000, 10000, 15),
    'exito_conexion': [True, True, False, True, True, True, False, False, True, True,
                       False, True, True,False,False],

    'Categoría Malware': ['Qbot malware', 'CryptoLocker ', 'Appearch', 'SQL Slammer','Keyloggers', 'Keyloggers','SQL Slammer','SQL Slammer','Qbot malware', 'CryptoLocker ',
                          'Qbot malware', 'CryptoLocker ', 'Appearch','Keyloggers', 'Keyloggers']                   
}
# Creamos el DataFrame
df = pd.DataFrame(data)
print("1. Explorando el DataFrame:")
print(df.head())
print("\nInformación del DataFrame:")
print(df.info())
print("\nEstadísticas descriptivas:")
print(df.describe())
```

```python
# 2. Selección de datos
print("\n2. Selección de datos:")
# Seleccionar una columna (devuelve una Serie)
print("\nSelección de una columna (Serie):")
print(df['protocolos inseguros vulnerables'])

# Seleccionar múltiples columnas (devuelve un DataFrame)
print("\nSelección de múltiples columnas (DataFrame):")
print(df[['ip_afectadas', 'protocolos inseguros vulnerables', 'puertos instrusivos']])

# Selección por índice con iloc
print("\nSelección por índice con iloc (primeras 2 filas, columnas 0 y 2):")
print(df.iloc[0:2, [0, 2]])

# Selección por nombre con loc
print("\nSelección por nombre con loc (filas donde protocolo es 'HTTP'):")
print(df.loc[df['protocolos inseguros vulnerables'] == 'HTTP', ['timestamp', 'ip_afectadas', 'puertos instrusivos']])
```

```python
# 3. Filtrado de datos
print("\n3. Filtrado de datos:")
# Filtrar conexiones exitosas
conexiones_exitosas = df[df['exito_conexion'] == True]
print("\nConexiones exitosas:")
print(conexiones_exitosas[['timestamp', 'ip_afectadas', 'protocolos inseguros vulnerables']])

# Filtrado múltiple (conexiones HTTP en puerto 80)
http_port80 = df[(df['protocolos inseguros vulnerables'] == 'HTTP') & (df['puertos instrusivos'] == 80)]
print("\nConexiones HTTP en puerto 80:")
print(http_port80)

# 4. Transformación de datos
print("\n4. Transformación de datos:")
# Añadir una nueva columna
df['hora'] = df['timestamp'].dt.hour
print("\nDataFrame con nueva columna 'hora':")
print(df[['timestamp', 'hora', 'protocolos inseguros vulnerables']])

# Aplicar una función con apply
def clasificar_protocolo(protocolo):
    if protocolo in ['SSH','RDP']:
        return 'ACCESO_REMOTO'
    elif protocolo == 'FTP':
        return 'TRANSFERENCIA_ARCHIVO'
    elif protocolo == 'HTTP':
        return 'WEB'
    else:
        return 'OTRO'

df['tipo_servicio'] = df['protocolos inseguros vulnerables'].apply(clasificar_protocolo)
print("\nDataFrame con clasificación de servicios:")
print(df[['protocolos inseguros vulnerables', 'tipo_servicio']])
```

```python
# 5. Agregación de datos
print("\n5. Agregación de datos:")
# Agrupar por protocolo y contar conexiones
conteo_por_protocolo = df.groupby('protocolos inseguros vulnerables').size()
print("\nConteo de conexiones por protocolo:")
print(conteo_por_protocolo)

# Estadísticas por protocolo
stats_por_protocolo = df.groupby('protocolos inseguros vulnerables')['bytes_enviados'].agg(['count', 'mean', 'min', 'max'])
print("\nEstadísticas de bytes enviados por protocolo:")
print(stats_por_protocolo)
```

```python
# 6. Pivotado de datos
print("\n6. Pivotado de datos:")
# Tabla pivote de bytes enviados por tipo de servicio y protocolo
pivot = pd.pivot_table(df, 
                      values='bytes_enviados', 
                      index='tipo_servicio', 
                      columns='exito_conexion', 
                      aggfunc='mean')
print("\nTabla pivote de bytes enviados por tipo de servicio y éxito de conexión:")
print(pivot)
```

```python
# 7. Detección de anomalías simples
print("\n7. Detección de anomalías simples:")
# Calcular el Z-score para identificar valores atípicos en bytes_enviados
df['z_score'] = (df['bytes_enviados'] - df['bytes_enviados'].mean()) / df['bytes_enviados'].std()
anomalias = df[abs(df['z_score']) > 1.5]  # Consideramos como anómalos Z-scores > 1.5
print("\nPosibles conexiones anómalas por volumen de datos transferidos:")
print(anomalias[['timestamp', 'ip_afectadas', 'protocolos inseguros vulnerables', 'bytes_enviados', 'z_score']])
```
